```python
def is_prime(num:int)-> bool:
    if num % 2 == 0:
        return False
    
    for i in range(2,num):
        if num % i == 0:
            return False
    
    return True

print("Twin Prime:")

for num in range(3,1000):
    if is_prime(num) and is_prime(num + 2):
        print(f" {num} and {num+2} ")
        



  
            
```

```python
def prime_factors(num:int)->list:
    factors=[]
    divisor=2
    while divisor * divisor <= num:
        
        while num % divisor ==0:
            factors.append(divisor)
            num //= divisor
            
        divisor += 1
    
    if num > 1:
        factors.append(num)
    
    return factors

prime_factors(25)
        
        
    
```

```python
#decimal to binary

def binary(num):
    binaries = []

    while num > 0:
        binaries.append(num % 2)
        num //= 2

    binaries.reverse()

    return binaries


print(binary(11))
```

```python
def is_perfect(num):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    return total == num


def perfect_numbers(start, end):
    for num in range(start, end + 1):
        if is_perfect(num):
            print(num)


perfect_numbers(1, 1000)
```
